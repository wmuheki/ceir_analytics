```python
import os
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Import and Combine all Fake, Genuine and Cloned Datasets

fake_dir = Path("/Volumes/E$/CEIR/Clean Dumps/Fake")
genuine_dir = Path("/Volumes/E$/CEIR/Clean Dumps/Genuine")
cloned_dir = Path("/Volumes/E$/CEIR/Clean Dumps/Cloned")

fake_df = pd.concat(
    (pd.read_parquet(f) for f in sorted(fake_dir.glob("*.parquet"))),
    ignore_index=True
)

genuine_df = pd.concat(
    (pd.read_parquet(f) for f in sorted(genuine_dir.glob("*.parquet"))),
    ignore_index=True
)


print(f"Fake: {len(fake_df):,} rows | Genuine: {len(genuine_df):,} rows")
```

```python
#drop uneccessary columns
fake_df = fake_df.drop(columns=['last_seen','imei_status','imsi','msisdn','rat','cgi','id_type','id_number','prefix','mno', 'gender', 'birth_year', 'age', 'district', 'mcc', 'country'])
genuine_df = genuine_df.drop(columns=['last_seen','imei_status','imsi','msisdn','rat','cgi','id_type','id_number','prefix','mno', 'gender', 'birth_year', 'age', 'district', 'mcc', 'country', 'tac', 'oem', 'brand','model', 'marketing_name',  'os_family', 'os_version', 'sim_slots', 'has_2g', 'has_3g', 'has_4g', 'has_5g', 'year_released'])
```

```python
fake_df = fake_df[
    (fake_df['imei_first_seen'] >= '2026-06-01') &
    (fake_df['imei_first_seen'] <  '2026-07-01')
].reset_index(drop=True)
```

```python
genuine_df = genuine_df[
    (genuine_df['imei_first_seen'] >= '2026-06-01') &
    (genuine_df['imei_first_seen'] <  '2026-07-01')
].reset_index(drop=True)
```
