### De Deduplicate - Consider doing this at the begining of ETL before enriching the data

```python
# 1. Ensure 'first_seen' is recognized as a datetime type (just in case it's currently a string)
df_domestic['first_seen'] = pd.to_datetime(df_domestic['first_seen'])

# 2. Sort the dataframe by 'first_seen' from oldest to newest (ascending)
df_sorted = df_domestic.sort_values(by='first_seen', ascending=True)

# 3. Drop duplicates, keeping only the 'first' row of each group (which will be the oldest due to sorting)
df_clean_domestic = df_sorted.drop_duplicates(subset=['imei', 'imsi'], keep='first')
```

```python
len(df_clean_domestic)
```

```python
# 1. Ensure 'first_seen' is recognized as a datetime type (just in case it's currently a string)
df_roamers['first_seen'] = pd.to_datetime(df_roamers['first_seen'])

# 2. Sort the dataframe by 'first_seen' from oldest to newest (ascending)
df_sorted_roamers = df_roamers.sort_values(by='first_seen', ascending=True)

# 3. Drop duplicates, keeping only the 'first' row of each group (which will be the oldest due to sorting)
df_clean_roamers = df_sorted_roamers.drop_duplicates(subset=['imei', 'imsi'], keep='first')
```
